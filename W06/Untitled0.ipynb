```python
# 載入資料科學必備套件
import numpy as np
import pandas as pd

# 設定亂數種子以確保結果可重現
random_seed = 42
rng = np.random.default_rng(random_seed)

print("套件載入成功！已建立亂數產生器 rng。")
```

```python
# 設定學生總人數
student_count = 60
n = student_count

# 1. 學生編號 (1 到 60)
student_ids = np.arange(1, n + 1)

# 2. 隨機選取科系
departments = rng.choice(["資訊管理", "企業管理", "金融管理"], size=n)

# 3. 睡眠時數 (平均 6.7，標準差 1.1，四捨五入小數點第一位)
sleep_hours = np.round(rng.normal(loc=6.7, scale=1.1, size=n), 1)

# 4. 螢幕使用時數 (平均 5.0，標準差 1.5，限制在 1.0~10.0 小時)
screen_time_hours = np.round(
    np.clip(rng.normal(loc=5.0, scale=1.5, size=n), 1.0, 10.0), 1
)

# 5. 讀書時數 (平均 2.5，標準差 1.0，限制在 0.5~6.0 小時)
study_hours = np.round(
    np.clip(rng.normal(loc=2.5, scale=1.0, size=n), 0.5, 6.0), 1
)

# 6. 出席率 (平均 88，標準差 8，限制在 60~100，整數)
attendance_rate = np.round(
    np.clip(rng.normal(loc=88, scale=8, size=n), 60, 100)
).astype(int)

# 組合成 pandas DataFrame
df = pd.DataFrame({
    "student_id": student_ids,
    "department": departments,
    "sleep_hours": sleep_hours,
    "screen_time_hours": screen_time_hours,
    "study_hours": study_hours,
    "attendance_rate": attendance_rate
})

# 顯示前五筆資料
df.head()
```

```python
# 產生隨機誤差項
random_error = rng.normal(loc=0, scale=5, size=n)

# 依照公式計算分數（分行呈現以利閱讀）
calculated_score = (
    50
    + (3 * df["study_hours"])
    + (2 * df["sleep_hours"])
    - (1.2 * df["screen_time_hours"])
    + (0.15 * df["attendance_rate"])
    + random_error
)

# 限制分數在 0~100 之間並轉為整數
df["quiz_score"] = np.round(np.clip(calculated_score, 0, 100)).astype(int)

# 選取特定欄位並顯示前五筆
selected_columns = [
    "student_id",
    "study_hours",
    "sleep_hours",
    "screen_time_hours",
    "attendance_rate",
    "quiz_score"
]

df[selected_columns]
```
